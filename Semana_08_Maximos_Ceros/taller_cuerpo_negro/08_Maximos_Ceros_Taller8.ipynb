## 1. Ceros de una función

Un **cero** o **raíz** de una función $f(x)$ es un valor $\alpha$ tal que

$$f(\alpha) = 0 .$$

Existen fórmulas exactas para unas pocas familias de funciones: la fórmula cuadrática para $ax^2 + bx + c$, o la factorización en casos particulares, como $x^3 - 6x^2 + 11x - 6 = (x-1)(x-2)(x-3)$. Para la mayoría de las ecuaciones que aparecen en física, como $x = 5(1 - e^{-x})$, no hay fórmula en términos de funciones elementales: la raíz se aproxima con un método iterativo, que parte de un valor inicial y produce una sucesión $x_0, x_1, x_2, \dots$ que se acerca a $\alpha$.

### 1.1 Localización de las raíces

Todos los métodos necesitan saber aproximadamente dónde está la raíz. El punto de partida es el **teorema del valor intermedio**: si $f$ es continua en $[a, b]$ y $f(a)\,f(b) < 0$, entonces $f$ tiene al menos un cero entre $a$ y $b$. El intervalo $[a, b]$ se llama **intervalo de arranque**.

En la práctica, los intervalos de arranque se encuentran evaluando $f$ en una malla y buscando los pares de puntos vecinos con signos distintos. Es una operación sobre arreglos, como las de la Semana 7. La ecuación de ejemplo de esta sección es

$$f(x) = x^6 - x - 1 = 0 ,$$

que tiene dos raíces reales: $\alpha_1 = -0.7780895987$ y $\alpha_2 = 1.134724138$.

```python
def f(x):
    """Función de ejemplo: f(x) = x**6 - x - 1."""
    return x**6 - x - 1

x = np.linspace(-1.5, 1.5, 31)            # malla gruesa, paso 0.1
y = f(x)

# Índices i tales que y[i] e y[i+1] tienen signos distintos
cambios = np.where(y[:-1]*y[1:] < 0)[0]
for i in cambios:
    print(f"cambio de signo en [{x[i]:5.2f}, {x[i+1]:5.2f}]:  f = {y[i]:8.4f}, {y[i+1]:8.4f}")

xf = np.linspace(-1.3, 1.3, 400)
fig, ax = plt.subplots(figsize=(7, 3.6))
ax.plot(xf, f(xf), color=AZUL)
ax.axhline(0, color=GRIS, lw=1)
ax.plot(x[cambios], 0*cambios, "o", color=DORADO, label="extremo izquierdo del intervalo")
ax.set_xlabel("$x$"); ax.set_ylabel("$f(x)$"); ax.set_ylim(-2, 4); ax.legend()
ax.set_title(r"$f(x) = x^6 - x - 1$")
plt.tight_layout(); plt.show()
```

<div style="border-left: 5px solid #B02A20; background-color: #FBEDEC; padding: 10px 14px; margin: 8px 0;">

**Evite** concluir que no hay raíz porque el barrido no encontró un cambio de signo. Una raíz doble, como la de $f(x) = (x-1)^2$, toca el eje sin cruzarlo, y dos raíces más cercanas que el paso de la malla producen dos cambios de signo que se cancelan entre puntos vecinos. Repita el barrido con una malla más fina y grafique la función; refinar no revela una raíz doble.

## 2. Método de bisección

El método parte de un intervalo $[a, b]$ con $f(a)\,f(b) < 0$. En cada iteración se calcula el punto medio $c = (a + b)/2$, se conserva la mitad del intervalo donde $f$ cambia de signo y se repite.

**Algoritmo**, con tolerancia $\varepsilon$ sobre el error absoluto de la raíz:

1. Verificar que $f(a)\,f(b) < 0$; si no se cumple, detenerse con un mensaje.
2. Calcular $c = (a + b)/2$.
3. Si $b - c \le \varepsilon$, aceptar $c$ como raíz y terminar.
4. Si $f(a)\,f(c) \le 0$, hacer $b = c$; en caso contrario, hacer $a = c$. Volver al paso 2.

**Cota del error.** Sea $\alpha$ la raíz y sean $a$, $b$ los extremos del intervalo inicial. Cada iteración divide el intervalo a la mitad, así que el punto medio $c_n$ de la iteración $n$ cumple

$$|\alpha - c_n| \le \frac{b - a}{2^n} ,$$

y el número de iteraciones necesario para una tolerancia $\varepsilon$ se conoce **antes** de ejecutar:

$$n \ge \frac{\log\left[(b - a)/\varepsilon\right]}{\log 2} .$$

Con $[a, b] = [1, 2]$ y $\varepsilon = 10^{-3}$, $n \ge 9.97$: 10 iteraciones.

El módulo siguiente contiene la función de ejemplo y su derivada. Los programas de las secciones 2 a 4 lo usan; para resolver otra ecuación basta cambiar el módulo.

```python
%%writefile modulo_ejemplo.f90
! ============================================================
! Módulo:    modulo_ejemplo
! Propósito: función de ejemplo f(x) = x**6 - x - 1 y su derivada.
! ============================================================
module modulo_ejemplo
  implicit none
  integer, parameter :: dp = kind(1.0d0)
  real(dp), parameter :: raiz_exacta = 1.134724138401519_dp   ! solo para medir errores
contains

  ! f: función cuyo cero se busca
  real(dp) function f(x)
    real(dp), intent(in) :: x
    f = x**6 - x - 1.0_dp
  end function f

  ! df: derivada analítica de f, f'(x) = 6 x**5 - 1
  real(dp) function df(x)
    real(dp), intent(in) :: x
    df = 6.0_dp*x**5 - 1.0_dp
  end function df

end module modulo_ejemplo
```

```python
%%writefile ejemplo_biseccion.f90
! ============================================================
! Programa:  ejemplo_biseccion
! Propósito: hallar la raíz de f en [1, 2] por bisección.
! Salida:    tabla en pantalla; errores_biseccion.dat
! ============================================================
program ejemplo_biseccion
  use modulo_ejemplo, only: dp, f, raiz_exacta
  implicit none
  real(dp), parameter :: tol = 1.0e-3_dp
  integer, parameter :: max_iter = 100      ! protección contra ciclos sin fin
  real(dp) :: a, b, c
  integer :: n, u

  a = 1.0_dp
  b = 2.0_dp
  ! --- Paso 1: el intervalo debe contener un cambio de signo
  if (f(a)*f(b) >= 0.0_dp) stop 'bisección: f(a) y f(b) no tienen signos distintos'

  open(newunit=u, file='errores_biseccion.dat', status='replace', action='write')
  print '(a)', '  n        a        b        c       b-c       f(c)'
  do n = 1, max_iter
     c = 0.5_dp*(a + b)                     ! paso 2: punto medio
     print '(i3,3f9.4,f10.5,f11.4)', n, a, b, c, b - c, f(c)
     write(u, '(i4,es14.6)') n, abs(c - raiz_exacta)
     if (b - c <= tol) exit                 ! paso 3: criterio de parada
     if (f(a)*f(c) <= 0.0_dp) then          ! paso 4: mitad con cambio de signo
        b = c
     else
        a = c
     end if
  end do
  close(u)
  print '(/,a,f8.4,a,i0,a)', 'Raíz aproximada: ', c, ' en ', n, ' iteraciones'
  print '(a,es9.2)', 'Error real:      ', abs(c - raiz_exacta)
end program ejemplo_biseccion
```

```python
!gfortran -fcheck=bounds -o ejemplo_biseccion modulo_ejemplo.f90 ejemplo_biseccion.f90 && ./ejemplo_biseccion
```

Después de 10 iteraciones, $b - c = 0.00098 < 0.001$, y el error real es $9.4\times10^{-4}$: la cota se cumple. La columna $f(c)$ muestra que la cercanía a la raíz no mejora de manera regular: en la iteración 9, $c = 1.1348$ estaba mucho más cerca que en la 10, con $f(c) = 0.0004$ frente a $-0.0096$.

**Propiedades del método**

- Si $f$ es continua y el intervalo de arranque es válido, **siempre converge**.
- El error garantizado se reduce a la mitad en cada iteración: la convergencia es **lineal**, con razón $1/2$. Se gana una cifra decimal cada 3.3 iteraciones aproximadamente.
- No aprovecha la forma de la función, solo el signo de sus valores.
- Encuentra una sola raíz por intervalo y no detecta las raíces de multiplicidad par, porque ahí $f$ no cambia de signo.

## 3. Método de Newton-Raphson

Se parte de una estimación $x_n$ de la raíz. La función se reemplaza por su recta tangente en $x_n$ y la siguiente estimación $x_{n+1}$ es el punto donde esa recta cruza el eje $x$. La recta tangente es

$$y = f(x_n) + f'(x_n)\,(x - x_n) ,$$

donde $f(x_n)$ es el valor de la función en la estimación actual y $f'(x_n)$ es su derivada, la pendiente de la tangente. Con $y = 0$:

$$x_{n+1} = x_n - \frac{f(x_n)}{f'(x_n)} .$$

**Error.** Sea $\alpha$ la raíz. La serie de Taylor de $f$ alrededor de $x_n$, evaluada en $\alpha$, es

$$0 = f(\alpha) = f(x_n) + (\alpha - x_n)\,f'(x_n) + \frac{1}{2}(\alpha - x_n)^2 f''(c_n) ,$$

con $c_n$ algún punto entre $\alpha$ y $x_n$. Dividiendo por $f'(x_n)$ y usando la fórmula de la iteración:

$$\alpha - x_{n+1} = (\alpha - x_n)^2\left[\frac{-f''(c_n)}{2 f'(x_n)}\right] .$$

El error de la iteración siguiente es proporcional al **cuadrado** del error actual: la convergencia es **cuadrática**, y el número de cifras correctas se duplica aproximadamente en cada paso, siempre que la estimación esté cerca de la raíz y $f'(\alpha) \neq 0$.

**Estimación del error.** La raíz no se conoce, pero cerca de ella $\alpha - x_n \approx -f(x_n)/f'(x_n) = x_{n+1} - x_n$. La diferencia entre dos iteraciones consecutivas mide el error de la anterior, y esa es la cantidad que se usa como criterio de parada.

```python
%%writefile ejemplo_newton.f90
! ============================================================
! Programa:  ejemplo_newton
! Propósito: hallar la raíz de f partiendo de x0 = 1.5 con el
!            método de Newton-Raphson.
! Salida:    tabla en pantalla; errores_newton.dat
! ============================================================
program ejemplo_newton
  use modulo_ejemplo, only: dp, f, df, raiz_exacta
  implicit none
  real(dp), parameter :: tol = 1.0e-10_dp
  integer, parameter :: max_iter = 50
  real(dp) :: x, paso, dfx
  integer :: n, u

  x = 1.5_dp                                ! estimación inicial x0
  open(newunit=u, file='errores_newton.dat', status='replace', action='write')
  write(u, '(i4,es14.6)') 0, abs(x - raiz_exacta)
  print '(a)', ' n       x_n              f(x_n)     x_n - x_(n-1)   alfa - x_(n-1)'
  print '(i2,f17.12,es13.3)', 0, x, f(x)
  do n = 1, max_iter
     dfx = df(x)
     if (abs(dfx) < tiny(1.0_dp)) stop 'Newton: derivada nula'
     paso = -f(x)/dfx                       ! corrección de Newton
     x = x + paso
     print '(i2,f17.12,es13.3,2es15.3)', n, x, f(x), paso, raiz_exacta - (x - paso)
     write(u, '(i4,es14.6)') n, abs(x - raiz_exacta)
     if (abs(paso) < tol) exit              ! dos iteraciones ya coinciden
  end do
  close(u)
  if (n > max_iter) print '(a)', 'Newton: no convergió en max_iter iteraciones'
  print '(a,es10.2)', 'Residuo final |f(x)|: ', abs(f(x))    ! revisión del resultado
end program ejemplo_newton
```

```python
!gfortran -fcheck=bounds -o ejemplo_newton modulo_ejemplo.f90 ejemplo_newton.f90 && ./ejemplo_newton
```

Las dos últimas columnas casi coinciden desde la cuarta iteración: $x_n - x_{n-1}$ estima bien el error de $x_{n-1}$. El error pasa de $4.7\times10^{-3}$ a $5.4\times10^{-5}$ y a $6.9\times10^{-9}$: el exponente se duplica. Comparado con bisección, Newton usa dos evaluaciones por iteración ($f$ y $f'$) y necesita conocer $f'$.

### 3.1 Dependencia del valor inicial

Newton no tiene garantía de convergencia. Si $x_0$ está lejos de la raíz, la tangente puede enviar la iteración a cualquier parte. El ejemplo clásico es

$$f(x) = \tan(\pi x) - x - 6 ,$$

que tiene una raíz en $\alpha = 0.45104726$ y un polo en $x = 0.5$, donde $\tan(\pi x)$ diverge. El programa siguiente parte de tres valores iniciales. La función se escribe en el mismo programa porque solo se usa aquí.

```python
%%writefile ejemplo_newton_falla.f90
! ============================================================
! Programa:  ejemplo_newton_falla
! Propósito: efecto del valor inicial en el método de Newton
!            para f(x) = tan(pi x) - x - 6.
! Salida:    tabla en pantalla
! ============================================================
program ejemplo_newton_falla
  implicit none
  integer, parameter :: dp = kind(1.0d0)
  real(dp), parameter :: pi = 4*atan(1.0_dp)
  real(dp) :: x(3)                          ! una estimación por columna
  integer :: n

  x = [0.48_dp, 0.40_dp, 0.01_dp]           ! tres valores iniciales
  print '(a)', ' n    x0 = 0.48     x0 = 0.40     x0 = 0.01'
  do n = 1, 11
     x = x - f(x)/df(x)                     ! una iteración de las tres columnas
     print '(i2,3f14.7)', n, x
  end do

contains

  elemental real(dp) function f(x)
    real(dp), intent(in) :: x
    f = tan(pi*x) - x - 6.0_dp
  end function f

  ! derivada analítica: pi/cos(pi x)**2 - 1
  elemental real(dp) function df(x)
    real(dp), intent(in) :: x
    df = pi/cos(pi*x)**2 - 1.0_dp
  end function df

end program ejemplo_newton_falla
```

```python
!gfortran -fcheck=bounds -o ejemplo_newton_falla ejemplo_newton_falla.f90 && ./ejemplo_newton_falla
```

Solo la primera columna converge a $0.45104726$. Las otras dos parten de una región donde la tangente cruza el eje lejos de la raíz: la sucesión salta entre ramas de la tangente y no se estabiliza, aunque cada iteración se calcula sin error de programación.

<div style="border-left: 5px solid #B02A20; background-color: #FBEDEC; padding: 10px 14px; margin: 8px 0;">

**Evite** usar Newton sin límite de iteraciones y sin revisar el resultado. El método puede diverger, saltar entre ramas o converger a una raíz distinta de la buscada. Fije siempre un máximo de iteraciones y compruebe que $|f(x)|$ es pequeño al terminar.

</div>

<div style="border-left: 5px solid #2E7D32; background-color: #EAF5EA; padding: 10px 14px; margin: 8px 0;">

**Consejo.** Unas pocas iteraciones de bisección mejoran el valor inicial de Newton, pero no garantizan su convergencia. Para conservar la garantía, un método híbrido mantiene el intervalo con cambio de signo y vuelve a bisección cuando el paso de Newton no es adecuado.

</div>

## 4. Método de la secante, orden de convergencia y criterios de parada

### 4.1 Método de la secante

Cuando $f'$ no se conoce o es costosa, la pendiente de la tangente se reemplaza por la de la recta secante que pasa por las dos últimas estimaciones:

$$x_{n+1} = x_n - f(x_n)\,\frac{x_n - x_{n-1}}{f(x_n) - f(x_{n-1})} ,$$

donde $x_n$ y $x_{n-1}$ son las dos últimas estimaciones y el cociente es el inverso de la pendiente de la secante. El método necesita dos valores iniciales, usa una sola evaluación nueva de $f$ por iteración y, para una raíz simple, su orden de convergencia es $p = (1 + \sqrt{5})/2 \approx 1.618$: más lento que Newton por iteración, más rápido que bisección y, cuando evaluar $f'$ cuesta tanto como evaluar $f$, más económico que Newton por evaluación.

```python
%%writefile ejercicio_secante.f90
! ============================================================
! Programa:  ejercicio_secante   (Ejercicio de calentamiento 1)
! Propósito: raíz de f con el método de la secante.
! Salida:    tabla en pantalla; errores_secante.dat
! ============================================================
program ejercicio_secante
  use modulo_ejemplo, only: dp, f, raiz_exacta
  implicit none
  real(dp), parameter :: tol = 1.0e-10_dp
  integer, parameter :: max_iter = 15
  real(dp) :: x0, x1, x_nuevo
  integer :: n, u

  x0 = 1.0_dp
  x1 = 2.0_dp
  open(newunit=u, file='errores_secante.dat', status='replace', action='write')
  print '(a)', ' n      x_n             error'
  do n = 2, max_iter
     if (f(x1) == f(x0)) stop 'secante: secante horizontal (denominador nulo)'
     ! la pendiente de la tangente se reemplaza por la de la secante (x0, x1)
     x_nuevo = x1 - f(x1)*(x1 - x0)/(f(x1) - f(x0))
     print '(i2,f17.12,es12.3)', n, x_nuevo, abs(x_nuevo - raiz_exacta)
     write(u, '(i4,es14.6)') n, abs(x_nuevo - raiz_exacta)
     if (abs(x_nuevo - x1) < tol) exit      ! dos estimaciones consecutivas coinciden
     x0 = x1                                ! la ventana avanza: se descarta la más vieja
     x1 = x_nuevo
  end do
  close(u)
end program ejercicio_secante
```

### 4.2 Orden de convergencia

Se dice que un método tiene **orden** $p$ si cerca de la raíz el error $e_n = |\alpha - x_n|$ cumple

$$e_{n+1} \approx C\,e_n^{\,p} ,$$

donde $C$ es una constante que depende de la función y $p$ es el orden: $p = 1$ para la cota de bisección, $p \approx 1.618$ en la secante y $p = 2$ en Newton. Tomando logaritmos en tres iteraciones consecutivas se obtiene $p$ sin conocer $C$:

$$p \approx \frac{\ln(e_{n+1}/e_n)}{\ln(e_n/e_{n-1})} .$$

La celda siguiente lee los archivos que escribieron los programas de bisección y Newton y, si ya se resolvió el Ejercicio 1, también el de la secante. El orden se estima con la última terna de errores que no esté dominada por el redondeo.

```python
import os

def orden(errores):
    """Estimación del orden p con la última terna de errores mayores que 1e-13."""
    e = errores[errores > 1e-13]
    return np.log(e[-1]/e[-2])/np.log(e[-2]/e[-3])

fig, ax = plt.subplots(figsize=(7, 4))
for nombre, archivo, color, marca in [("bisección", "errores_biseccion.dat", AZUL, "o"),
                                      ("secante", "errores_secante.dat", VERDE, "^"),
                                      ("Newton", "errores_newton.dat", DORADO, "s")]:
    if not os.path.exists(archivo) or os.path.getsize(archivo) == 0:
        print(f"{nombre:>10}: sin datos todavía")
        continue
    n, e = np.loadtxt(archivo, unpack=True)
    if np.any(np.isnan(e)) or e[-1] > 1e-3 and nombre != "bisección":
        print(f"{nombre:>10}: sin convergencia todavía")
        continue
    ax.semilogy(n, np.maximum(e, 1e-17), marca + "-", ms=4, color=color, label=nombre)
    if nombre != "bisección":                  # ver el recuadro siguiente
        print(f"{nombre:>10}: orden estimado p = {orden(e):.2f}")
ax.set_xlabel("iteración $n$"); ax.set_ylabel(r"error $|\alpha - x_n|$")
ax.set_title(r"Convergencia para $x^6 - x - 1 = 0$"); ax.legend()
plt.tight_layout(); plt.show()
```

<div style="border-left: 5px solid #B02A20; background-color: #FBEDEC; padding: 10px 14px; margin: 8px 0;">

**Evite** medir el orden de bisección con errores consecutivos: su error real oscila, porque el punto medio puede caer muy cerca de la raíz por casualidad. Lo que se reduce a la mitad en cada iteración es la **cota** $(b-a)/2^n$, no el error real.

</div>

### 4.3 Criterios de parada

Un método iterativo necesita decidir cuándo terminar sin conocer la raíz. Los criterios habituales son:

| Criterio | Condición | Qué controla |
|---|---|---|
| Intervalo | $b - c \le \varepsilon$ | Error garantizado en bisección |
| Paso | $\lvert x_{n+1} - x_n\rvert < \varepsilon$ | Error estimado en Newton y secante |
| Residuo | $\lvert f(x_n)\rvert < \varepsilon$ | Qué tan cerca de cero está $f$ |
| Iteraciones | $n \ge n_{\max}$ | Evita un ciclo sin fin |

Un residuo pequeño no implica un error pequeño: si $f$ es casi plana cerca de la raíz, $f(x)$ puede ser diminuto lejos de $\alpha$. Un paso pequeño tampoco asegura el éxito. Por esa razón se usa el criterio de paso o de intervalo y se revisa el residuo al final.

## 5. Máximos y mínimos como ceros de la derivada (Python)

En un máximo o en un mínimo interior de una función suave, la derivada es cero. Los ceros de $f'$ son los candidatos a extremos:

$$f'(x^*) = 0, \qquad f''(x^*) < 0 \ \ \text{(máximo)}, \qquad f''(x^*) > 0 \ \ \text{(mínimo)},$$

donde $x^*$ es la posición del extremo. Estas condiciones sobre $f''$ son suficientes, no necesarias: $f' = 0$ no distingue entre un máximo y un mínimo, y si $f''(x^*) = 0$ el criterio no decide. Por ejemplo, $x^4$ tiene un mínimo en cero y $x^3$ no tiene extremo allí.

El ejemplo de esta sección es una función proporcional a la distribución de rapideces de Maxwell-Boltzmann en unidades reducidas:

$$f(v) = v^2\,e^{-v^2/2} ,$$

donde $v$ es la rapidez, medida en unidades de $\sqrt{k_B T/m}$, con $k_B$ la constante de Boltzmann, $T$ la temperatura y $m$ la masa de la molécula. El factor $v^2$ cuenta los estados con rapidez $v$ y el exponencial es el factor de Boltzmann.

### 5.1 Máximo analítico

Derivando,

$$f'(v) = \left(2v - v^3\right)e^{-v^2/2} = 0 \quad\Longrightarrow\quad v^* = \sqrt{2} = 1.414214 ,$$

que es la **rapidez más probable**. Este valor sirve de referencia para medir los errores de los métodos numéricos.

### 5.2 Máximo sobre la malla y cambio de signo de la derivada numérica

Con datos en una malla de paso $h$ hay tres aproximaciones, de menor a mayor esfuerzo:

1. **Mayor valor del arreglo**, con `np.argmax`. Su error es de orden $h$ y no se reduce de manera regular al refinar la malla.
2. **Cero lineal de la derivada numérica.** Con la derivada central $d_i$, se localiza el par de puntos donde $d$ pasa de positiva a negativa y se interpola linealmente entre ellos.
3. **Cero del spline de la derivada numérica** (sección 5.3).

```python
def maxwell(v):
    """Distribución de Maxwell-Boltzmann en unidades reducidas."""
    return v**2*np.exp(-v**2/2)

def derivada_central(y, h):
    """Derivada central en los puntos interiores de una malla de paso h."""
    return (y[2:] - y[:-2])/(2*h)

def cero_lineal(x, d):
    """Interpola linealmente el primer paso de d de positiva a negativa."""
    i = np.where((d[:-1] > 0) & (d[1:] <= 0))[0][0]
    return x[i] - d[i]*(x[i+1] - x[i])/(d[i+1] - d[i])

h = 0.2
v = np.arange(0.0, 6.0 + h/2, h)     # malla de la distribución
y = maxwell(v)
d = derivada_central(y, h)           # derivada en v[1:-1]

v_ref = np.sqrt(2)
print(f"máximo del arreglo: v = {v[np.argmax(y)]:.6f},  error = {abs(v[np.argmax(y)] - v_ref):.2e}")
v_lin = cero_lineal(v[1:-1], d)
print(f"cero lineal de d:   v = {v_lin:.6f},  error = {abs(v_lin - v_ref):.2e}")
```

### 5.3 Spline cúbico de la derivada numérica

Un **spline cúbico** es una función formada por un polinomio de grado 3 en cada intervalo de la malla, con valor, primera y segunda derivada continuas en los puntos de unión. Pasa exactamente por los datos y da una función definida entre los puntos de la malla. Construirlo requiere resolver un sistema tridiagonal, tema del Capítulo 5; por eso se usa la clase `CubicSpline` de SciPy, que se llama igual que una función y además da sus derivadas y sus raíces:

| Expresión | Resultado |
|---|---|
| `s = CubicSpline(x, y)` | spline que pasa por los puntos $(x_i, y_i)$ |
| `s(x0)` | valor del spline en `x0` |
| `s(x0, 1)` | primera derivada del spline en `x0` |
| `s.derivative()` | nuevo spline: la derivada de `s` |
| `s.roots(extrapolate=False)` | ceros de `s` dentro del intervalo de los datos |

El método del cronograma aplica el spline a la derivada numérica $d_i$: el cero del spline es el candidato a máximo, y la derivada del spline en ese punto aproxima $f''$, cuyo signo indica si es un máximo.

```python
vi = v[1:-1]                           # puntos donde existe la derivada central
s_d = CubicSpline(vi, d)               # spline de la derivada numérica

candidatos = s_d.roots(extrapolate=False)
for r in candidatos:
    tipo = "máximo" if s_d(r, 1) < 0 else "mínimo"
    print(f"cero en v = {r:.6f}: f'' aprox. = {s_d(r, 1):8.4f} -> {tipo}")
v_spl = candidatos[s_d(candidatos, 1) < 0][0]

fig, ax = plt.subplots(2, 1, figsize=(7, 5), sharex=True)
vf = np.linspace(0, 4, 400)
ax[0].plot(vf, maxwell(vf), color=GRIS, lw=3, alpha=0.5, label="$f(v)$ exacta")
ax[0].plot(v, y, "o", ms=4, color=AZUL, label="malla, $h = 0.2$")
ax[0].axvline(v_ref, color=DORADO, ls="--", label=r"$v^* = \sqrt{2}$")
ax[0].set_xlim(0, 4); ax[0].set_ylabel("$f(v)$"); ax[0].legend()
ax[1].plot(vf[1:], s_d(vf[1:]), color=VERDE, label="spline de la derivada")
ax[1].plot(vi, d, "o", ms=4, color=AZUL, label="derivada central")
ax[1].axhline(0, color=GRIS, lw=1); ax[1].axvline(v_ref, color=DORADO, ls="--")
ax[1].set_xlabel("$v$"); ax[1].set_ylabel("$f'(v)$"); ax[1].legend()
plt.tight_layout(); plt.show()
```

### 5.4 Precisión al refinar la malla

La tabla compara las tres aproximaciones con mallas cada vez más finas, y agrega una cuarta: el spline de la función misma, cuya derivada se obtiene con `s.derivative()`, sin pasar por la derivada central.

```python
def errores_maximo(h):
    """Errores de las cuatro aproximaciones del máximo con paso h."""
    v = np.arange(0.0, 6.0 + h/2, h)
    y = maxwell(v)
    d = derivada_central(y, h)
    s_d = CubicSpline(v[1:-1], d)                       # spline de la derivada
    r = s_d.roots(extrapolate=False)
    e_spl_d = abs(r[s_d(r, 1) < 0][0] - v_ref)
    ds = CubicSpline(v, y).derivative()                 # derivada del spline de f
    r = ds.roots(extrapolate=False)
    r = r[(ds(r, 1) < 0) & (r > 0.5)]
    return [abs(v[np.argmax(y)] - v_ref), abs(cero_lineal(v[1:-1], d) - v_ref),
            e_spl_d, abs(r[0] - v_ref)]

pasos = [0.4, 0.2, 0.1, 0.05]
tabla = np.array([errores_maximo(h) for h in pasos])
print("    h     arreglo   lineal    spline de d   spline de f")
for h, fila in zip(pasos, tabla):
    print(f"{h:6.2f} " + "".join(f"{e:11.2e}" for e in fila))
print("\nrazón entre errores consecutivos")
print(np.round(tabla[:-1]/tabla[1:], 2)[:, 1:])
```

El mayor valor del arreglo no mejora de manera regular: su error depende de dónde cae la malla. El cero lineal y el spline de la derivada mejoran con razón $\approx 4$, es decir, con orden 2, aunque un spline cúbico interpola con mucho mejor orden: el error lo fija la **derivada central**, de orden 2, que alimenta al spline. Un método es tan preciso como su eslabón más débil. El spline de la función misma evita ese eslabón y tiene errores mucho menores.

## 6. Taller 8: la radiación de cuerpo negro


### A. Enunciado y requisitos

La radiancia espectral de un cuerpo negro a temperatura $T$ está dada por la **ley de Planck**:

$$B_\lambda(\lambda, T) = \frac{2hc^2}{\lambda^5}\,\frac{1}{e^{hc/(\lambda k_B T)} - 1} ,$$

donde $B_\lambda$ es la potencia emitida por unidad de área, de ángulo sólido y de longitud de onda [W m⁻² sr⁻¹ m⁻¹], $\lambda$ la longitud de onda [m], $T$ la temperatura [K], $h = 6.62607015\times10^{-34}$ J s la constante de Planck, $c = 299792458$ m/s la rapidez de la luz en el vacío y $k_B = 1.380649\times10^{-23}$ J/K la constante de Boltzmann. El factor $2hc^2/\lambda^5$ cuenta los modos electromagnéticos con su energía y el factor con el exponencial es la ocupación de esos modos. Estas tres constantes son exactas en el Sistema Internacional.

**Variable adimensional.** Con $x = hc/(\lambda k_B T)$, la razón entre la energía de un fotón y la energía térmica, el espectro queda

$$B_\lambda \propto \frac{x^5}{e^x - 1} .$$

**Ley de desplazamiento de Wien.** El máximo del espectro cumple $dB_\lambda/dx = 0$, que se reduce a

$$x = n\left(1 - e^{-x}\right), \qquad n = 5 ,$$

ecuación trascendente sin solución en funciones elementales, con la raíz trivial $x = 0$ y una raíz positiva $x_5$. Esa raíz fija la constante de Wien $b$:

$$\lambda_{\max}\,T = b = \frac{hc}{k_B\,x_5} .$$

Si el espectro se expresa por unidad de frecuencia, $B_\nu \propto x^3/(e^x - 1)$, y el máximo cumple la misma ecuación con $n = 3$.

**Ley de Stefan-Boltzmann.** La potencia total emitida por unidad de área es $M = \int_0^\infty \pi B_\lambda\,d\lambda = \sigma T^4$, con

$$\sigma = \frac{2\pi k_B^4}{h^3 c^2}\,I, \qquad I = \int_0^\infty \frac{x^3}{e^x - 1}\,dx ,$$

donde $\sigma$ es la constante de Stefan-Boltzmann [W m⁻² K⁻⁴] e $I$ una integral pura, cuyo valor exacto es $\pi^4/15$. Es una integral impropia. El cambio de variable $x = t/(1 - t)$, con $dx = dt/(1-t)^2$, la lleva al intervalo finito $[0, 1]$:

$$I = \int_0^1 \frac{x(t)^3}{e^{x(t)} - 1}\,\frac{dt}{(1 - t)^2} .$$

**Parte B: código.** Con $[a,b]=[1,10]$ y $\varepsilon=10^{-10}$, la cota de la sección 2 predice $n \ge \log(9/10^{-10})/\log 2 = 36.4$, es decir **37 iteraciones** de bisección; el programa imprime esa predicción para compararla con el conteo real.

```python
%%writefile modulo_ceros.f90
! ============================================================
! Módulo:    modulo_ceros
! Propósito: ecuación de Wien F(x) = x - n(1 - e^{-x}), su derivada
!            y tres métodos para hallar ceros: bisección, Newton y
!            secante. Cada método cuenta sus iteraciones y, si se le
!            pasa una unidad de archivo abierta (u_reg /= 0), registra
!            cada valor nuevo para analizar la convergencia. Se compara
!            con /= 0 y no con > 0 porque open(newunit=...) entrega
!            números de unidad NEGATIVOS.
! Convención del registro: una fila por iteración con
!            (método, número de iteración, x), numerada desde 1.
! ============================================================
module modulo_ceros
  implicit none
  integer, parameter :: dp = kind(1.0d0)
  integer, parameter :: max_iter = 100      ! requisito 3 del taller

contains

  ! F: función cuyo cero se busca. n es el exponente del espectro
  !    (5 para B_lambda, 3 para B_nu); entra con intent(in) para que
  !    ningún método pueda modificarlo por accidente.
  real(dp) function F(x, n)
    real(dp), intent(in) :: x
    integer,  intent(in) :: n
    F = x - real(n, dp)*(1.0_dp - exp(-x))
  end function F

  ! dF: derivada analítica, F'(x) = 1 - n e^{-x}. Se anula en x = ln(n),
  !     punto que Newton debe evitar (denominador nulo).
  real(dp) function dF(x, n)
    real(dp), intent(in) :: x
    integer,  intent(in) :: n
    dF = 1.0_dp - real(n, dp)*exp(-x)
  end function dF

  ! ----------------------------------------------------------
  ! Bisección (sección 2 del notebook).
  ! Entrada: n, intervalo [a0, b0], tolerancia tol, unidad u_reg
  !          (0 = no registrar).
  ! Salida:  raiz = último punto medio, iter = iteraciones usadas.
  ! a0 y b0 se copian a variables locales para no alterar los datos
  ! iniciales del programa principal.
  ! ----------------------------------------------------------
  subroutine biseccion(n, a0, b0, tol, u_reg, raiz, iter)
    integer,  intent(in)  :: n, u_reg
    real(dp), intent(in)  :: a0, b0, tol
    real(dp), intent(out) :: raiz
    integer,  intent(out) :: iter
    real(dp) :: a, b, c
    integer  :: k

    a = a0
    b = b0
    ! Paso 1: sin cambio de signo no hay garantía de raíz en [a, b]
    if (F(a, n)*F(b, n) >= 0.0_dp) stop 'biseccion: F(a) y F(b) no tienen signos distintos'

    do k = 1, max_iter
       c = 0.5_dp*(a + b)                          ! paso 2: punto medio
       if (u_reg /= 0) write(u_reg, '(i3,i5,es24.16)') 1, k, c
       if (b - c <= tol) exit                      ! paso 3: error garantizado <= tol
       if (F(a, n)*F(c, n) <= 0.0_dp) then         ! paso 4: conservar la mitad
          b = c                                    !         con cambio de signo
       else
          a = c
       end if
    end do
    if (k > max_iter) then                         ! el do terminó sin exit
       print '(a)', 'biseccion: no convergió en max_iter iteraciones'
       k = max_iter
    end if
    raiz = c
    iter = k
  end subroutine biseccion

  ! ----------------------------------------------------------
  ! Newton-Raphson (sección 3): x_{k+1} = x_k - F(x_k)/F'(x_k).
  ! Parada: |x_{k+1} - x_k| < tol, que estima el error de x_k.
  ! ----------------------------------------------------------
  subroutine newton(n, x0, tol, u_reg, raiz, iter)
    integer,  intent(in)  :: n, u_reg
    real(dp), intent(in)  :: x0, tol
    real(dp), intent(out) :: raiz
    integer,  intent(out) :: iter
    real(dp) :: x, paso, pendiente
    integer  :: k

    x = x0
    do k = 1, max_iter
       pendiente = dF(x, n)
       if (pendiente == 0.0_dp) stop 'newton: derivada nula, la tangente no corta el eje'
       paso = -F(x, n)/pendiente                   ! corrección de Newton
       x = x + paso
       if (u_reg /= 0) write(u_reg, '(i3,i5,es24.16)') 2, k, x
       if (abs(paso) < tol) exit
    end do
    if (k > max_iter) then
       print '(a)', 'newton: no convergió en max_iter iteraciones'
       k = max_iter
    end if
    raiz = x
    iter = k
  end subroutine newton

  ! ----------------------------------------------------------
  ! Secante (sección 4.1): la derivada se reemplaza por la pendiente
  ! de la recta que pasa por las dos últimas estimaciones.
  ! La iteración k produce x_{k+1}; la primera produce x_2.
  ! Se guarda F(x_anterior) para hacer una sola evaluación nueva por paso.
  ! ----------------------------------------------------------
  subroutine secante(n, x0, x1, tol, u_reg, raiz, iter)
    integer,  intent(in)  :: n, u_reg
    real(dp), intent(in)  :: x0, x1, tol
    real(dp), intent(out) :: raiz
    integer,  intent(out) :: iter
    real(dp) :: x_ant, x_act, x_nuevo, f_ant, f_act, denominador
    integer  :: k

    x_ant = x0
    x_act = x1
    f_ant = F(x_ant, n)
    f_act = F(x_act, n)
    do k = 1, max_iter
       denominador = f_act - f_ant
       if (denominador == 0.0_dp) stop 'secante: F(x_k) = F(x_{k-1}), secante horizontal'
       x_nuevo = x_act - f_act*(x_act - x_ant)/denominador
       if (u_reg /= 0) write(u_reg, '(i3,i5,es24.16)') 3, k, x_nuevo
       ! desplazar la ventana de dos puntos antes de la prueba de parada
       x_ant = x_act
       f_ant = f_act
       x_act = x_nuevo
       f_act = F(x_act, n)
       if (abs(x_act - x_ant) < tol) exit
    end do
    if (k > max_iter) then
       print '(a)', 'secante: no convergió en max_iter iteraciones'
       k = max_iter
    end if
    raiz = x_act
    iter = k
  end subroutine secante

end module modulo_ceros
```

```python
%%writefile programa_wien.f90
! ============================================================
! Programa:  programa_wien
! Propósito: hallar x5 y x3 (raíces positivas de x = n(1 - e^{-x}))
!            con bisección, Newton y secante; repetir Newton desde
!            x0 = 0.5 y calcular la constante de Wien b = hc/(kB x5).
! Salida:    resultado_wien.dat, iteraciones_wien.dat, constante_wien.dat
! ============================================================
program programa_wien
  use modulo_ceros, only: dp, F, biseccion, newton, secante
  implicit none
  ! Constantes exactas del SI (2019)
  real(dp), parameter :: h_planck = 6.62607015e-34_dp    ! J s
  real(dp), parameter :: c_luz    = 299792458.0_dp       ! m/s
  real(dp), parameter :: k_boltz  = 1.380649e-23_dp      ! J/K
  real(dp), parameter :: tol = 1.0e-10_dp
  integer,  parameter :: exponentes(2) = [5, 3]          ! n = 5 (B_lambda), n = 3 (B_nu)

  real(dp) :: raiz(3, 2)          ! raiz(método, caso de n)
  integer  :: iter(3, 2)          ! iteraciones(método, caso de n)
  real(dp) :: x_trivial, b_wien
  integer  :: it_trivial, j, m, n, u_res, u_it, u_reg, u_cte
  integer  :: n_predicho

  ! Iteraciones que predice la cota de bisección (sección 2):
  ! n >= log((b-a)/eps)/log 2 con [1, 10] y eps = 1e-10
  n_predicho = ceiling(log(9.0_dp/tol)/log(2.0_dp))
  print '(a,i0)', 'Bisección: iteraciones predichas por la cota = ', n_predicho

  open(newunit=u_it, file='iteraciones_wien.dat', status='replace', action='write')

  do j = 1, 2
     n = exponentes(j)
     ! solo se registran las iteraciones del caso n = 5
     u_reg = 0
     if (n == 5) u_reg = u_it
     call biseccion(n, 1.0_dp, 10.0_dp, tol, u_reg, raiz(1, j), iter(1, j))
     call newton   (n, 10.0_dp,         tol, u_reg, raiz(2, j), iter(2, j))
     call secante  (n, 10.0_dp, 9.0_dp, tol, u_reg, raiz(3, j), iter(3, j))
  end do
  close(u_it)

  ! Newton desde x0 = 0.5: a la izquierda del punto donde F' = 0 (x = ln 5)
  call newton(5, 0.5_dp, tol, 0, x_trivial, it_trivial)

  ! Constante de Wien con x5 de Newton
  b_wien = h_planck*c_luz/(k_boltz*raiz(2, 1))

  ! --- resultado_wien.dat: n, método, raíz, iteraciones
  open(newunit=u_res, file='resultado_wien.dat', status='replace', action='write')
  do j = 1, 2
     do m = 1, 3
        write(u_res, '(i3,i3,es24.16,i5)') exponentes(j), m, raiz(m, j), iter(m, j)
     end do
  end do
  write(u_res, '(i3,i3,es24.16,i5)') 5, 4, x_trivial, it_trivial
  close(u_res)

  ! --- constante_wien.dat: x5, x3, b [m K]
  open(newunit=u_cte, file='constante_wien.dat', status='replace', action='write')
  write(u_cte, '(3es24.16)') raiz(2, 1), raiz(2, 2), b_wien
  close(u_cte)

  ! --- resumen en pantalla, con el residuo como revisión final
  print '(/,a)', '  n  método            raíz           iter    |F(raíz)|'
  do j = 1, 2
     do m = 1, 3
        print '(i3,i5,f22.14,i7,es13.3)', exponentes(j), m, raiz(m, j), iter(m, j), &
              abs(F(raiz(m, j), exponentes(j)))
     end do
  end do
  print '(i3,a,f22.14,i7,es13.3)', 5, '  N(0.5)', x_trivial, it_trivial, abs(F(x_trivial, 5))
  print '(/,a,es16.9,a)', 'Constante de Wien b = ', b_wien, ' m K'
end program programa_wien
```

```python
%%writefile modulo_romberg.f90
! ============================================================
! Módulo:    modulo_romberg
! Propósito: integración de Romberg (tabla triangular con
!            extrapolación de Richardson) con conteo de
!            evaluaciones, y los integrandos del Taller 7:
!              opción 1 -> 2*gamma*v(t)**2   (energía disipada, Parte B)
!              opción 2 -> 4/(1 + x**2)      (valor de pi, Parte D)
!            Se usa sin cambios en las Partes B y D.
!            Taller 8 (Parte C): se AGREGAN dos integrandos para la
!            ley de Stefan-Boltzmann; romberg y su criterio de parada
!            no se modificaron.
!              opción 3 -> x**3/(e**x - 1), x = t/(1-t), por 1/(1-t)**2
!              opción 4 -> x**3/(e**x - 1), sin cambio de variable
! ============================================================
module modulo_romberg
  implicit none
  integer,  parameter :: dp = kind(1.0d0)
  real(dp), parameter :: pi = 4*atan(1.0_dp)

  ! Parámetros del oscilador amortiguado x(t) = exp(-gamma t) cos(omega t)
  real(dp), parameter :: gamma = 0.5_dp           ! amortiguamiento [1/s]
  real(dp), parameter :: omega = 2*pi             ! frecuencia angular [rad/s]
  real(dp), parameter :: omega0_2 = omega**2 + gamma**2   ! omega_0^2 [1/s^2]

  ! Selección del integrando
  integer, parameter :: INTEG_ENERGIA = 1
  integer, parameter :: INTEG_PI      = 2
  integer, parameter :: INTEG_PLANCK_T = 3      ! Taller 8, con cambio de variable
  integer, parameter :: INTEG_PLANCK   = 4      ! Taller 8, sin cambio de variable

  ! Taller 8: por encima de x_corte el integrando de Planck es < 1e-250
  ! y e**x se acerca al desborde de doble precisión (x ~ 709)
  real(dp), parameter :: x_corte = 600.0_dp

contains

  ! ----------------------------------------------------------
  ! posicion: x(t) del oscilador amortiguado [m]
  ! ----------------------------------------------------------
  elemental real(dp) function posicion(t)
    real(dp), intent(in) :: t
    posicion = exp(-gamma*t)*cos(omega*t)
  end function posicion

  ! ----------------------------------------------------------
  ! velocidad: v(t) exacta del oscilador [m/s]
  ! ----------------------------------------------------------
  elemental real(dp) function velocidad(t)
    real(dp), intent(in) :: t
    velocidad = -exp(-gamma*t)*(gamma*cos(omega*t) + omega*sin(omega*t))
  end function velocidad

  ! ----------------------------------------------------------
  ! energia: E(t) = v^2/2 + omega0^2 x^2/2, por unidad de masa [J/kg]
  ! ----------------------------------------------------------
  elemental real(dp) function energia(t)
    real(dp), intent(in) :: t
    energia = 0.5_dp*velocidad(t)**2 + 0.5_dp*omega0_2*posicion(t)**2
  end function energia

  ! ----------------------------------------------------------
  ! planck: x^3/(e^x - 1)                              (Taller 8)
  ! Límites programados aparte:
  !   x -> 0:      x^3/(e^x - 1) ~ x^2 -> 0 (evita el 0/0 en x = 0)
  !   x > x_corte: integrando despreciable (evita el desborde de e^x)
  ! ----------------------------------------------------------
  elemental real(dp) function planck(x)
    real(dp), intent(in) :: x
    if (x <= 0.0_dp .or. x > x_corte) then
       planck = 0.0_dp
    else
       planck = x**3/(exp(x) - 1.0_dp)
    end if
  end function planck

  ! ----------------------------------------------------------
  ! planck_t: integrando con x = t/(1-t), dx = dt/(1-t)^2   (Taller 8)
  ! Lleva la integral impropia en [0, inf) al intervalo [0, 1].
  !   t = 0 -> x = 0, integrando 0
  !   t = 1 -> x y 1/(1-t)^2 divergen, pero e^{-x} gana: límite 0
  ! ----------------------------------------------------------
  elemental real(dp) function planck_t(t)
    real(dp), intent(in) :: t
    real(dp) :: x
    if (t <= 0.0_dp .or. t >= 1.0_dp) then
       planck_t = 0.0_dp
    else
       x = t/(1.0_dp - t)
       planck_t = planck(x)/(1.0_dp - t)**2
    end if
  end function planck_t

  ! ----------------------------------------------------------
  ! integrando: integrando elegido por 'opcion' (elemental:
  ! acepta un arreglo de puntos)
  ! Entradas: opcion (INTEG_ENERGIA, INTEG_PI, INTEG_PLANCK_T o
  !           INTEG_PLANCK), x
  ! Salida:   f(x); con una opción desconocida devuelve 0
  ! ----------------------------------------------------------
  elemental real(dp) function integrando(opcion, x)
    integer,  intent(in) :: opcion
    real(dp), intent(in) :: x
    select case (opcion)
    case (INTEG_ENERGIA)
       integrando = 2*gamma*velocidad(x)**2      ! potencia disipada dE/dt con signo +
    case (INTEG_PI)
       integrando = 4/(1 + x**2)
    case (INTEG_PLANCK_T)
       integrando = planck_t(x)
    case (INTEG_PLANCK)
       integrando = planck(x)
    case default
       integrando = 0.0_dp
    end select
  end function integrando

  ! ----------------------------------------------------------
  ! romberg: integral de 'opcion' en [a, b] por la tabla de Romberg
  ! Entradas: opcion, a, b, tol (tolerancia del criterio de la
  !           diagonal), kmax (nivel máximo)
  ! Salidas:  S(0:kmax, 0:kmax), tabla triangular (S(k,j) con j<=k);
  !           k_final, último nivel calculado;
  !           n_eval, evaluaciones del integrando (2**k_final + 1);
  !           convergio, .true. si |S(k,k) - S(k-1,k-1)| < tol
  ! ----------------------------------------------------------
  subroutine romberg(opcion, a, b, tol, kmax, S, k_final, n_eval, convergio)
    integer,  intent(in)  :: opcion, kmax
    real(dp), intent(in)  :: a, b, tol
    real(dp), allocatable, intent(out) :: S(:,:)
    integer,  intent(out) :: k_final, n_eval
    logical,  intent(out) :: convergio
    real(dp) :: h, suma, factor
    integer  :: k, j, i, m

    ! Tabla con índices desde 0, como en las fórmulas
    allocate(S(0:kmax, 0:kmax))
    S = 0.0_dp

    ! Nivel 0: un solo trapecio sobre todo el intervalo
    h = b - a
    S(0,0) = 0.5_dp*h*(integrando(opcion, a) + integrando(opcion, b))
    n_eval = 2
    convergio = .false.
    k_final = kmax

    do k = 1, kmax
       h = 0.5_dp*h
       ! Solo los puntos nuevos (puntos medios del nivel anterior):
       ! x = a + (2i - 1) h, i = 1, ..., 2**(k-1)
       m = 2**(k - 1)
       suma = sum(integrando(opcion, [( a + (2*i - 1)*h, i = 1, m )]))
       n_eval = n_eval + m
       S(k,0) = 0.5_dp*S(k-1,0) + h*suma

       ! Extrapolación de Richardson (4**j en real: evita desborde)
       do j = 1, k
          factor = 4.0_dp**j
          S(k,j) = (factor*S(k,j-1) - S(k-1,j-1))/(factor - 1)
       end do

       ! Criterio de parada: dos elementos diagonales consecutivos
       if (abs(S(k,k) - S(k-1,k-1)) < tol) then
          k_final = k
          convergio = .true.
          exit
       end if
    end do
  end subroutine romberg

end module modulo_romberg
```

```python
%%writefile programa_sigma.f90
! ============================================================
! Programa:  programa_sigma
! Propósito: calcular I = int_0^inf x^3/(e^x - 1) dx con el módulo
!            de Romberg del Taller 7:
!            (1) con el cambio de variable x = t/(1-t) en [0, 1] y
!            (2) sin cambio de variable en [0, 100];
!            obtener sigma = 2 pi kB^4/(h^3 c^2) I y comparar con pi^4/15.
! Salida:    resultado_sigma.dat (k_final, evaluaciones, I, sigma)
! ============================================================
program programa_sigma
  use modulo_romberg, only: dp, pi, romberg, INTEG_PLANCK_T, INTEG_PLANCK
  implicit none
  real(dp), parameter :: h_planck = 6.62607015e-34_dp    ! J s
  real(dp), parameter :: c_luz    = 299792458.0_dp       ! m/s
  real(dp), parameter :: k_boltz  = 1.380649e-23_dp      ! J/K
  real(dp), parameter :: tol = 1.0e-8_dp
  integer,  parameter :: kmax = 20                       ! límite de niveles
  real(dp), allocatable :: S(:, :)                       ! tabla de Romberg (la reserva romberg)
  real(dp) :: factor, integral(2), sigma(2), I_exacta
  integer  :: k_fin(2), n_ev(2), fila, u
  logical  :: convergio(2)

  factor   = 2.0_dp*pi*k_boltz**4/(h_planck**3*c_luz**2)
  I_exacta = pi**4/15.0_dp

  ! (1) con cambio de variable, t en [0, 1]
  call romberg(INTEG_PLANCK_T, 0.0_dp, 1.0_dp, tol, kmax, S, k_fin(1), n_ev(1), convergio(1))
  integral(1) = S(k_fin(1), k_fin(1))
  ! (2) sin cambio de variable, x en [0, 100]
  !     (S es intent(out) allocatable: se libera y se reserva de nuevo)
  call romberg(INTEG_PLANCK, 0.0_dp, 100.0_dp, tol, kmax, S, k_fin(2), n_ev(2), convergio(2))
  integral(2) = S(k_fin(2), k_fin(2))
  sigma = factor*integral

  do fila = 1, 2
     if (.not. convergio(fila)) print '(a,i0,a)', 'fila ', fila, ': Romberg no convergió en kmax niveles'
  end do

  open(newunit=u, file='resultado_sigma.dat', status='replace', action='write')
  do fila = 1, 2
     write(u, '(i4,i10,2es24.16)') k_fin(fila), n_ev(fila), integral(fila), sigma(fila)
  end do
  close(u)

  print '(a)', ' fila  k_final  evaluaciones        I                 error |I - pi^4/15|'
  do fila = 1, 2
     print '(i4,i9,i12,es24.15,es14.3)', fila, k_fin(fila), n_ev(fila), integral(fila), &
           abs(integral(fila) - I_exacta)
  end do
  print '(/,a,es24.15)', 'pi^4/15             = ', I_exacta
  print '(a,es16.9,a)',  'sigma (con cambio)  = ', sigma(1), ' W m^-2 K^-4'
end program programa_sigma
```

```python
%%writefile maximo_espectro.py
# ============================================================
# Programa:  maximo_espectro.py   (Taller 8, Parte D)
# Propósito: hallar la longitud de onda del máximo del espectro de
#            un cuerpo negro a T = 5772 K (modelo del Sol) en mallas
#            de paso 100, 50, 25 y 12.5 nm, de tres maneras, y medir
#            el error frente a la ley de Wien, lambda_max = b/T.
#   (a) mayor valor del arreglo
#   (b) cero del spline de la derivada central (derivada programada)
#   (c) cero de la derivada del spline de B_lambda
# Entrada:   constante_wien.dat (x5, x3, b) de la Parte B
# Salida:    maximo_espectro.dat, espectro_solar.png
# Restricción del taller: no se usa np.gradient ni scipy.optimize;
#            CubicSpline (.roots y .derivative) solo en esta parte.
# ============================================================
import numpy as np
import matplotlib
matplotlib.use("Agg")                    # guarda la figura sin ventana
import matplotlib.pyplot as plt
from scipy.interpolate import CubicSpline

# Constantes exactas del SI
H, C, KB = 6.62607015e-34, 299792458.0, 1.380649e-23
T_SOL = 5772.0                           # K
PASOS = [100.0, 50.0, 25.0, 12.5]        # nm, en el orden que pide el enunciado
AZUL, DORADO, ROJO, GRIS = "#162A54", "#C9972C", "#B02A20", "#7F7F7F"


def planck_lambda(lam_nm, T):
    """Radiancia espectral B_lambda [W m^-2 sr^-1 m^-1] con lambda en nm."""
    lam = lam_nm*1e-9                                  # nm -> m
    x = H*C/(lam*KB*T)                                 # energía del fotón / kB T
    return 2*H*C**2/lam**5/np.expm1(x)                 # expm1 = e^x - 1 sin cancelación


def derivada_central(y, h):
    """Derivada central de orden 2 en los puntos interiores (sección 5.2)."""
    return (y[2:] - y[:-2])/(2*h)


def elegir_maximo(spline_derivada, referencia):
    """Ceros de la derivada con segunda derivada negativa (máximos).

    En una malla gruesa el spline oscila y produce ceros espurios; si
    queda más de un máximo se conserva el más cercano a 'referencia',
    el resultado del método (a).
    """
    r = spline_derivada.roots(extrapolate=False)
    r = r[np.isfinite(r)]
    r = r[spline_derivada(r, 1) < 0]                   # f'' < 0: máximo
    if len(r) == 0:
        raise RuntimeError("no se encontró un máximo en la malla")
    return r[np.argmin(np.abs(r - referencia))]


def maximos(paso):
    """Las tres estimaciones de lambda_max [nm] para una malla de paso dado."""
    lam = np.arange(100.0, 3000.0 + paso/2, paso)      # + paso/2 incluye 3000 nm
    B = planck_lambda(lam, T_SOL)
    lam_a = lam[np.argmax(B)]                                          # (a)
    s_d = CubicSpline(lam[1:-1], derivada_central(B, paso))            # (b)
    lam_b = elegir_maximo(s_d, lam_a)
    dB = CubicSpline(lam, B).derivative()                              # (c)
    lam_c = elegir_maximo(dB, lam_a)
    return lam, B, (lam_a, lam_b, lam_c)


# --- Constantes de la Parte B (Fortran)
x5, x3, b_wien = np.loadtxt("constante_wien.dat")
lam_ref = b_wien/T_SOL*1e9                             # ley de Wien, nm
lam_nu = H*C/(KB*T_SOL*x3)*1e9                         # máximo de B_nu, expresado en lambda
print(f"lambda_max = b/T = {lam_ref:.4f} nm;  máximo en frecuencia -> lambda = {lam_nu:.1f} nm\n")

# --- Errores absolutos en las cuatro mallas
filas = []
print(" paso[nm]  lambda(a)    error(a)   lambda(b)    error(b)   lambda(c)    error(c)")
for paso in PASOS:
    lam, B, estimaciones = maximos(paso)
    errores = [abs(e - lam_ref) for e in estimaciones]
    filas.append([paso, *errores])
    print(f"{paso:8.1f}" + "".join(f"{e:11.4f}{err:12.3e}" for e, err in zip(estimaciones, errores)))
    if paso == 25.0:
        lam25, B25, est25 = lam, B, estimaciones       # malla de la figura

filas = np.array(filas)
np.savetxt("maximo_espectro.dat", filas, fmt="%24.16e")
print("\nrazón entre errores consecutivos (b) y (c):")
print(np.round(filas[:-1, 2:]/filas[1:, 2:], 2))

# --- Figura: espectro con la malla de 25 nm
fig, ax = plt.subplots(figsize=(7.5, 4.2), dpi=130)
lam_fino = np.linspace(100, 3000, 2000)
ax.plot(lam_fino, planck_lambda(lam_fino, T_SOL)*1e-9, color=GRIS, lw=2.5, alpha=0.5,
        label="Planck exacta")
ax.plot(lam25, B25*1e-9, "o", ms=3, color=AZUL, label=r"malla $\Delta\lambda = 25$ nm")
ax.axvline(lam_ref, color=DORADO, ls="--",
           label=rf"$\lambda_{{\max}} = b/T = {lam_ref:.1f}$ nm")
ax.plot(est25[2], planck_lambda(est25[2], T_SOL)*1e-9, "*", ms=12, color=DORADO,
        label=rf"máximo con spline (c): {est25[2]:.1f} nm")
ax.axvline(lam_nu, color=ROJO, ls=":",
           label=rf"máximo en frecuencia: $hc/(k_BTx_3) = {lam_nu:.0f}$ nm")
ax.set_xlabel(r"$\lambda$ [nm]")
ax.set_ylabel(r"$B_\lambda$ [W m$^{-2}$ sr$^{-1}$ nm$^{-1}$]")
ax.set_title(r"Espectro de cuerpo negro a $T = 5772$ K (Sol)")
ax.set_xlim(100, 3000); ax.grid(alpha=0.3); ax.legend(fontsize=8)
fig.tight_layout()
fig.savefig("espectro_solar.png")
print("\nfigura guardada en espectro_solar.png")
```

## E. Respuestas de interpretación

Los números salen de `resultado_wien.dat`, `iteraciones_wien.dat`, `constante_wien.dat`, `resultado_sigma.dat` y `maximo_espectro.dat`.

**1. Costo de los métodos.** Con $n = 5$ y $\varepsilon = 10^{-10}$, los tres métodos llegan a $x_5 = 4.9651142317$:

| Método | Iteraciones | Evaluaciones por iteración |
|---|---|---|
| Bisección, $[1, 10]$ | 37 | 1 ($F$) |
| Newton, $x_0 = 10$ | 4 | 2 ($F$ y $F'$) |
| Secante, $x_0 = 10$, $x_1 = 9$ | 5 | 1 ($F$) |

Bisección hace exactamente las 37 iteraciones que predice la cota $n \ge \log(9/10^{-10})/\log 2 = 36.4$. Con los errores $e_k = |x_5 - x_k|$ de `iteraciones_wien.dat` y $p \approx \ln(e_{k+1}/e_k)/\ln(e_k/e_{k-1})$:

- **Newton:** errores $3.4\times10^{-2}$, $2.0\times10^{-5}$, $7.1\times10^{-12}$ → $p \approx 2.00$ (cuadrático).
- **Secante:** errores $3.3\times10^{-2}$, $8.6\times10^{-4}$, $5.0\times10^{-7}$, $7.8\times10^{-12}$ → las dos últimas ternas dan $p \approx 2.05$ y $1.49$; con tan pocas iteraciones la estimación oscila alrededor del valor teórico $1.618$.

Por evaluaciones de función, Newton usa $\approx 8$ y la secante $\approx 7$: aquí la secante es tan económica como Newton sin necesitar $F'$, y ambos son unas 5 veces más baratos que bisección.

**2. Valor inicial.** Desde $x_0 = 0.5$, Newton llega en 6 iteraciones a $x \approx 9\times10^{-17}$, es decir, a la **raíz trivial $x = 0$**. Es una raíz correcta porque $F(0) = 0 - 5(1 - 1) = 0$. La causa es la forma de $F$: $F'(x) = 1 - 5e^{-x}$ se anula en $x = \ln 5 \approx 1.61$. A la izquierda de ese punto la pendiente es negativa y la tangente cruza el eje hacia $x = 0$, no hacia $x_5$. No es la respuesta física porque $x = hc/(\lambda k_B T) = 0$ corresponde a $\lambda \to \infty$, donde $B_\lambda \to 0$: la raíz trivial aparece al dividir $dB_\lambda/dx = 0$ entre el factor $x^4$ y no es el máximo del espectro. Para obtener $x_5$ hay que partir a la derecha de $\ln 5$ o usar primero bisección en un intervalo que excluya el cero, como $[1, 10]$.

**3. Parada prematura.** Sin cambio de variable, en $[0, 100]$ Romberg se detiene en el nivel $k = 1$ con 3 evaluaciones e $I \approx 1.6\times10^{-15}$, frente a $\pi^4/15 = 6.4939$. Los puntos que evalúa son $x = 0, 100$ (nivel 0) y $x = 50$ (nivel 1), donde el integrando vale $0$, $3.7\times10^{-38}$ y $2.4\times10^{-17}$. Toda el área está en $x \lesssim 20$ (el máximo del integrando está en $x_3 = 2.82$), y la malla gruesa no tiene ningún punto allí. Por eso $R_{0,0} \approx 2\times10^{-36}$ y $R_{1,1} \approx 1.6\times10^{-15}$ difieren en menos de $10^{-8}$: el criterio de la diagonal compara dos números casi nulos y se cumple **sin que la integral haya convergido**. El criterio es absoluto y no sabe que la escala de la respuesta es $\sim 6$.

**Comprobación propuesta:** usar un criterio **relativo**, $|R_{k,k} - R_{k-1,k-1}| < \varepsilon\,|R_{k,k}|$, junto con un nivel mínimo antes de aceptar. En una prueba aparte, con criterio relativo y $k \ge 4$, el mismo Romberg en $[0, 100]$ se detiene en $k = 10$ con $I = 6.4939394023$ (error $9\times10^{-12}$). También sirve comparar el resultado con una cota física o graficar el integrando en los puntos evaluados. El cambio de variable resuelve el problema de raíz: con $t \in [0, 1]$, Romberg converge en $k = 10$ (1025 evaluaciones) con error $2.7\times10^{-13}$ y da $\sigma = 5.670374419\times10^{-8}$ W m⁻² K⁻⁴.

**4. Eslabón más débil.** Al reducir $\Delta\lambda$ a la mitad ($100 \to 50 \to 25 \to 12.5$ nm):

| $\Delta\lambda$ [nm] | error (a) | error (b) | error (c) |
|---|---|---|---|
| 100 | 2.04 | 12.9 | 1.20 |
| 50 | 2.04 | 3.39 | $3.2\times10^{-2}$ |
| 25 | 2.04 | 0.85 | $2.9\times10^{-3}$ |
| 12.5 | 2.04 | 0.21 | $3.1\times10^{-4}$ |

- **(b)** baja con razones $3.8,\ 3.98,\ 4.00$: orden 2. Lo limita la **derivada central**, de error $O(\Delta\lambda^2)$, que alimenta al spline. El spline interpola con mejor orden, pero no puede corregir los datos que recibe.
- **(c)** baja con razones $37,\ 11,\ 9.3$, cerca de 8 o más (orden $\gtrsim 3$, el de la derivada de un spline cúbico), y con $\Delta\lambda = 12.5$ nm su error es casi 700 veces menor que el de (b).
- **(a)** queda fijo en $2.04$ nm porque todas las mallas contienen $\lambda = 500$ nm, que en todas es el punto de la malla más cercano a $\lambda_{\max} = 502.04$ nm (en la malla más fina, el siguiente punto, 512.5 nm, queda a 10.5 nm). El máximo del arreglo solo puede devolver un punto de la malla; su error baja únicamente cuando aparece un punto más cercano, lo que exigiría un paso de unos 3 nm.

**5. Dos máximos.** $B_\lambda$ y $B_\nu$ son **densidades**: potencia por unidad de $\lambda$ y por unidad de $\nu$. Como $\nu = c/\lambda$, $B_\lambda = B_\nu\,|d\nu/d\lambda| = B_\nu\,c/\lambda^2$, y el factor $c/\lambda^2$ desplaza el máximo. En longitud de onda el pico está en $b/T = 502$ nm (verde-azulado); en frecuencia, en $\lambda = hc/(k_B T x_3) = 883$ nm (infrarrojo cercano). Ninguno de los dos es "el" color del Sol: la posición del máximo de una densidad depende de la variable que se use para repartir la potencia, y no es una propiedad física del espectro. Lo que no cambia con la variable es la potencia en una banda, $\int B_\lambda\,d\lambda = \int B_\nu\,d\nu$. El color que vemos lo da el espectro completo ponderado por la respuesta del ojo, y para el Sol fuera de la atmósfera es prácticamente **blanco**.
